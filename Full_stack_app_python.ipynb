```python
print("Welcome to Student Task Manager")
print("1. Add a Task")
print("2. View Tasks")
print("3. Exit")
```

```python
tasks = []

task = input("Enter your task: ")
tasks.append(task)

print("Task added successfully!")
print("Your tasks:", tasks)
```

```python
print("My Tasks:")
for task in tasks:
    print("-", task)
```

```python
def add_task():
    task = input("Enter a new task: ")
    tasks.append(task)
    print("Task added successfully!")

add_task()
print("All Tasks:", tasks)
```

```python
def delete_task():
    print("Your Tasks:", tasks)
    task = input("Enter the task to delete: ")

    if task in tasks:
        tasks.remove(task)
        print("Task deleted successfully!")
    else:
        print("Task not found!")

delete_task()
print("Remaining Tasks:", tasks)
```

```python
completed_tasks = []

def complete_task():
    print("Your Tasks:", tasks)
    task = input("Enter the completed task: ")

    if task in tasks:
        completed_tasks.append(task)
        print("Task marked as completed!")
    else:
        print("Task not found!")

complete_task()

print("Completed Tasks:", completed_tasks)
```
